## Finish class _isu_directory_

- Finish `__init__`
- Create `display()` method.
  - You can choose how exactly to display the information but it should look "nice" and be informative; for example someone who executes the code in your notebook should be able to clearly understand what the strings or numbers represents.
  - For example, a disembodied number such as 555102201 is not great; some text that associates it with the fact that this is an ID number is very helpful.


```python
class isu_directory:
    def __init__(self, name, isu_id, dob):
        self.name = name
        self.isu_id = isu_id
        self.dob = dob

    def display(self):
        print(f"Name: {self.name}")
        print(f"ISU ID: {self.isu_id}")
        print(f"Date of Birth: {self.dob}")
```
