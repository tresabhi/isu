Quiz 3 (I think?) toy neural network. I start off with importing Pytorch and the neural network submodule:


```python
import torch
import torch.nn as nn
```

And then the class that extends from the nn.Module class:


```python
# Parenthesis here indicate inheritance.
class Model(nn.Module):

    # Double underscore init double underscore is the constructor.
    def __init__(self):

        # Initializing the parent class.
        super(Model, self).__init__()

        # Each of the layers, all connected up. The first one of course has only
        # 2 and the last has an output of 1.
        self.fc1 = nn.Linear(2, 10)
        self.fc2 = nn.Linear(10, 10)
        self.fc3 = nn.Linear(10, 1)

    # Propagation through the layers.
    def forward(self, x):

        # Propagation uses the sigmoid activation function.
        x = torch.sigmoid(self.fc1(x))
        x = torch.sigmoid(self.fc2(x))
        x = torch.sigmoid(self.fc3(x))
        return x
```

Time to use it:


```python
# An instance of the model class:
model = Model()

# The data provided:
x = torch.tensor([1.9, 4.9])
```

And the output:


```python
# In Python you can call an instance of a class like a function lol but
# that's how Pytorch does it.
output = model(x)

# Print.
print("Model output:", output.item())
```
